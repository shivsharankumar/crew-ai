```python
from crewai import LLM
llm =LLM(
    model="openai/gpt-4o",
    temperature=0.7
)
llm.call("What is the capital of France?")
```

```python
original_email = """ 
Looping in priya .TAS and PRX updates are in the deck.ETA for SDS integration is friday.
Let's sync up tomorrow if SYNCBOT allows.ping me if any blockers.
"""
#lets now build a tools  who converts the internal organization abbreviations to full form and then we will use that tool in our agent to rewrite the email in a more professional way.
from crewai.tools import BaseTool
class ReplaceJargonsTool(BaseTool):
    name:str="ReplaceJargonsTool"
    description:str="A tool to replace internal organization abbreviations to full form"
    def _run(self,email:str)->str:
        jargon_map = {
            "TAS": "Technical Architecture Specification",
            "PRX": "Project Phoenix (Internal AI revamp Project)",
            "SDS": "smart data syncer",
            "SYNCBOT": "Synchronization Bot",
            "DBX": "clinet database cluster",
            "WIP": "Work In Progress",
            "POC": "Proof of Concept",
            "ping": "contact",
        }
        suggestions=[]
        email_lower=email.lower()
        for jargon,replacement in jargon_map.items():
            if jargon.lower() in email_lower:
                suggestions.append(f"Replace {jargon} with {replacement}")
        
        return "\n".join(suggestions) if suggestions else "No jargons found"

jt=ReplaceJargonsTool()
jt.run(original_email)
```

```python
from crewai import Agent, Task,Crew
email_assistant = Agent(
    name="Email Assistant",
    role="email_Assistant",
    goal="Improve emails and make them sound professional and clear",
    backstory="A highly experience communication expert skilled in professional email writing",
    tools=[jt],
    verbose=True,
    llm=llm
)
#2.specify task

email_task=Task(
    description=f"""Take the following rough email and rewrite it into a professional and polished version.
    Expand abbreviations:
    '''{original_email}'''""",
    agent=email_assistant,
    expected_output="A polished and professional email that effectively communicates the message and is free of grammatical errors with proper formatting and content."   

)
crew=Crew(
    agents=[email_assistant],
    tasks=[email_task],
    verbose=True
)
result=crew.kickoff()
print("-----------------Result-------------------------------\n",result)
```
