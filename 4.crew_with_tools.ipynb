```python
from crewai import LLM
llm =LLM(
    model="openai/gpt-4o",
    temperature=0.7
)
llm.call("What is the capital of France?")
```

```python
tool = SerperDevTool()
result = tool._run(search_query="Latest AI news")
print(result)


```

```python
from crewai import Agent,Task,Crew
research_agent = Agent(
    role="Research Specialist",
    goal="research interesting facts about the topic: {topic}",
    # goal="Research and analyze data for the team",
    backstory="You are an expert at finding relevant and factual data.",
    verbose=True,
    tools=[SerperDevTool()],
    llm=llm
)
writer_agent = Agent(
    role="Creative Writer",
    goal="Write a short blog summary using the research",
    backstory="A highly experienced communication expert skilled in professional writing and summarizing",
    verbose=True,
    llm=llm
)
```

```python
task1=Task(
    description="find 3-5 interesting and recent facts about {topic} as of year 2026",
    expected_output="A bullet list of 3-5 facts",
    agent=research_agent
)
task2=Task(
    description="Write a 100 word blog post summary about  using the research",
    expected_output="A short blog summary",
    agent=writer_agent,
    context=[task1]
)
crew=Crew(
    agents=[research_agent,writer_agent],
    tasks=[task1,task2],
    verbose=True
)
crew.kickoff(inputs={"topic":"The Future of Electric Vehicles"})
```
